1. Start Spark Session & Load Data

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("EmployeeOperations").getOrCreate()

# Example dataset
data = [
    ("Aman", 28, 40000, "Delhi"),
    ("Ravi", 35, 60000, "Mumbai"),
    ("Neha", 32, 55000, "Delhi"),
    ("Sita", 29, 70000, "Bangalore"),
    ("Karan", 40, 80000, "Delhi")
]

columns = ["name", "age", "salary", "city"]

df = spark.createDataFrame(data, columns)

df.show()
```

2. Filter employees with salary > 50,000

```python
df.filter(df.salary > 50000).show()
```

3. Filter employees whose age > 30

```python
df.filter(df.age > 30).show()
```

4. Count total number of employees

```python
df.count()
```

5. Show employees from a specific city (Delhi)

```python
df.filter(df.city == "Delhi").show()
```

6 Sample Dataset (with department & gender)

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import count, sum, avg, max

spark = SparkSession.builder.appName("GroupByOperations").getOrCreate()

data = [
    ("Aman", 28, 40000, "IT", "Male", "Delhi"),
    ("Ravi", 35, 60000, "HR", "Male", "Mumbai"),
    ("Neha", 32, 55000, "IT", "Female", "Delhi"),
    ("Sita", 29, 70000, "Finance", "Female", "Bangalore"),
    ("Karan", 40, 80000, "IT", "Male", "Delhi"),
    ("Pooja", 27, 45000, "HR", "Female", "Mumbai")
]

columns = ["name", "age", "salary", "department", "gender", "city"]

df = spark.createDataFrame(data, columns)

df.show()
```

 7Group by department & count employees

```python
df.groupBy("department").agg(count("*").alias("employee_count")).show()
```

 8 Total salary per department

```python
df.groupBy("department").agg(sum("salary").alias("total_salary")).show()
```

 9 Average salary per department

```python
df.groupBy("department").agg(avg("salary").alias("avg_salary")).show()
```

 10 Group by city & find max salary

```python
df.groupBy("city").agg(max("salary").alias("max_salary")).show()
```

 11 Group by department & gender & count employees

```python
df.groupBy("department", "gender") \
  .agg(count("*").alias("employee_count")) \
  .show()
```

Bonus: Multiple aggregations together

```python
df.groupBy("department").agg(
    count("*").alias("count"),
    sum("salary").alias("total_salary"),
    avg("salary").alias("avg_salary")
).show()
```

 12 Sample Dataset (with role added)

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum, avg, max, min, count, countDistinct, desc

spark = SparkSession.builder.appName("MultiAggregation").getOrCreate()

data = [
    ("Aman", "IT", "Developer", 40000),
    ("Ravi", "HR", "Manager", 60000),
    ("Neha", "IT", "Developer", 55000),
    ("Sita", "Finance", "Analyst", 70000),
    ("Karan", "IT", "Manager", 80000),
    ("Pooja", "HR", "Executive", 45000)
]

columns = ["name", "department", "role", "salary"]

df = spark.createDataFrame(data, columns)
df.show()
```

 13 Multiple aggregations on salary per department

```python
df.groupBy("department").agg(
    sum("salary").alias("total_salary"),
    avg("salary").alias("average_salary"),
    max("salary").alias("max_salary"),
    min("salary").alias("min_salary")
).show()
```

 14 Group by department & role

```python
df.groupBy("department", "role").agg(
    sum("salary").alias("total_salary"),
    avg("salary").alias("average_salary"),
    count("*").alias("employee_count")
).show()
```

 15 Count distinct roles in each department

```python
df.groupBy("department").agg(
    countDistinct("role").alias("distinct_roles")
).show()
```

Top 3 departments with highest total salary

```python
df.groupBy("department").agg(
    sum("salary").alias("total_salary")
).orderBy(desc("total_salary")).limit(3).show()
```

Bonus (Alternative sorting way)

```python
df.groupBy("department") \
  .agg(sum("salary").alias("total_salary")) \
  .sort(desc("total_salary")) \
  .show(3)
```
