```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, DoubleType, IntegerType

# Initialize Spark Session (if not already running)
spark = SparkSession.builder.appName("MockDataGeneration").getOrCreate()

# 1. Define the structural schema explicitly using PySpark types
schema = StructType([
    StructField("first_name", StringType(), True),
    StructField("last_name", StringType(), True),
    StructField("job_title", StringType(), True),
    StructField("dob", StringType(), True),
    StructField("email", StringType(), True),
    StructField("phone", StringType(), True),
    StructField("salary", DoubleType(), True),
    StructField("department_id", IntegerType(), True)
])

# 2. Define sample mock data rows
data = [
    ("John", "Doe", "Data Engineer", "1990-05-15", "john.doe@example.com", "555-0101", 85000.00, 1),
    ("Jane", "Smith", "Data Scientist", "1988-11-23", "jane.smith@example.com", "555-0102", 95000.50, 1),
    ("Michael", "Brown", "Software Engineer", "1993-02-02", "michael.b@example.com", "555-0103", 80000.00, 2),
    ("Emily", "Davis", "Product Manager", "1985-08-30", "emily.d@example.com", "555-0104", 105000.00, 3),
    ("James", "Wilson", "HR Specialist", "1991-04-12", "james.w@example.com", "555-0105", 62000.00, 4),
    ("Sarah", "Johnson", "Data Engineer", "1994-07-19", "sarah.j@example.com", "555-0106", 87000.00, 1),
    ("David", "Martinez", "DevOps Engineer", "1989-12-05", "david.m@example.com", "555-0107", 92000.75, 2)
]

# 3. Create the DataFrame
emp_df = spark.createDataFrame(data, schema=schema)

# 4. Display the results
emp_df.show(truncate=False)
emp_df.printSchema()
```

```python

# ------------------------------------------------------------
# 5. Define Department Schema
# ------------------------------------------------------------
_dept_schema = """
department_id int,
department_name string,
description string,
city string,
state string,
country string
"""
```

```python
# ------------------------------------------------------------
# 6. Create Department DataFrame from Mock Data
# ------------------------------------------------------------

data_dept = [
    (1, "Engineering", "Software development", "San Francisco", "CA", "USA"),
    (2, "DevOps", "Infrastructure management", "New York", "NY", "USA"),
    (3, "Product", "Product management", "Seattle", "WA", "USA"),
    (4, "Human Resources", "HR management", "Austin", "TX", "USA")
]

dept = spark.createDataFrame(data_dept, schema=_dept_schema)

dept.show(truncate=False)
dept.printSchema()
```

```python
# 10. Create Temporary Views
# ------------------------------------------------------------
emp_df.createOrReplaceTempView("emp_view")
dept.createOrReplaceTempView("dept_view")
```

```python
# 11. Filter Employee Data Using SQL
# ------------------------------------------------------------
emp_filtered = spark.sql("""
    SELECT *
    FROM emp_view
    WHERE department_id = 1
""")
```

```python
# 12. Show Filtered Employee Data
# ------------------------------------------------------------
emp_filtered.show(truncate=False)
emp_filtered.printSchema()
```

```python
# 12. Show Filtered Data
# ------------------------------------------------------------
emp_filtered.show()
```

```python
# 13. Join Filtered Employee Data with Department Data
# ------------------------------------------------------------
emp_dept_joined = emp_filtered.join(dept, on="department_id", how="inner")

emp_dept_joined.show(truncate=False)
emp_dept_joined.printSchema()
```

```python
# 13. Extract Year from DOB
# ------------------------------------------------------------
emp_temp = spark.sql("""
    SELECT
        e.*,
        date_format(dob, 'yyyy') AS dob_year
    FROM emp_view e
""")
```

```python
# 14. Calculate Average Salary by Job Title in Filtered Department
# ------------------------------------------------------------
from pyspark.sql.functions import avg

avg_salary_by_job = emp_dept_joined.groupBy("job_title").agg(avg("salary").alias("average_salary"))

avg_salary_by_job.show()
avg_salary_by_job.printSchema()
```

```python
# 14. Create Another Temporary View
# ------------------------------------------------------------
emp_temp.createOrReplaceTempView("emp_temp_view")
```

```python
# 15. Display Updated Data
# ------------------------------------------------------------
spark.sql("SELECT * FROM emp_temp_view").show()
```

```python
# 16. Join Employee and Department Data
# ------------------------------------------------------------
emp_final = spark.sql("""
    SELECT /*+ BROADCAST(d) */
        e.*,
        d.department_name
    FROM emp_view e
    LEFT OUTER JOIN dept_view d
    ON e.department_id = d.department_id
""")
```

```python
# 17. Display Final Joined Data
# ------------------------------------------------------------
emp_final.show()
```

```python
# 18. Save DataFrame as Spark SQL Table
# ------------------------------------------------------------
emp_final.write.format("parquet").mode("overwrite").saveAsTable("emp_final")
```

```python
# 19. Read Data from Saved Table
# ------------------------------------------------------------
emp_new = spark.sql("SELECT * FROM emp_final")
```

```python
# 20. Show Table Data
# ------------------------------------------------------------
emp_new.show()
```
